# Customer Segmentation: Interpretation & Growth Actions

This notebook interprets the customer segments identified in Notebook 03 and translates them into concrete, business-actionable strategies.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load customer features saved from notebook 03
customer_features = pd.read_csv(
    Path("../data/processed/customer_features.csv")
)
```

```python
segment_summary = (
    customer_features
    .groupby("segment")
    .agg(
        customers=("CustomerID", "nunique"),
        avg_recency_days=("recency_days", "mean"),
        avg_frequency=("frequency", "mean"),
        avg_monetary_value=("monetary_value", "mean"),
        avg_order_value=("avg_order_value", "mean"),
    )
    .sort_values("avg_monetary_value", ascending=False)
)

segment_summary
```

```python
segment_names = {
    4: "High-Value Loyal",
    0: "Growing Repeat",
    3: "Bulk / Wholesale",
    2: "Low-Value / One-Time",
    1: "At-Risk / Churned",
}

segment_summary_named = (
    segment_summary
    .assign(segment_name=lambda df: df.index.map(segment_names))
    .reset_index()
)

segment_summary_named
```

## Segment: High-Value Loyal Customers

**Profile**
- Most recent purchasers (low recency)
- Very high purchase frequency
- Highest total customer lifetime value
- Moderate-to-high average order value

These customers represent the core revenue base and are critical to business stability.

**Business interpretation**
This segment consists of loyal repeat customers who purchase frequently and contribute disproportionately to revenue. They are highly engaged and have demonstrated trust in the brand.

**Recommended actions**
- Do **not** discount aggressively (protect margin)
- Offer early access to new products
- Provide priority fulfillment or exclusive bundles
- Align inventory planning to protect availability for this group

**Goal**
Retention and satisfaction. Losing customers in this segment has outsized negative impact.

## Segment: Growing Repeat Customers

**Profile**
- Recent purchasers
- Moderate purchase frequency
- Meaningful but not yet high total value
- Lower average order value than top-tier customers

This segment is large and represents the primary growth opportunity.

**Business interpretation**
These customers are transitioning from casual buyers to loyal repeat customers. Small improvements in retention or basket size can materially increase revenue.

**Recommended actions**
- Bundled product offerings
- Loyalty programs or reorder reminders
- Free-shipping thresholds to increase basket size
- Light, targeted promotions (avoid blanket discounts)

**Goal**
Move customers from this segment into the High-Value Loyal segment.

## Segment: Bulk / Wholesale Buyers

**Profile**
- Infrequent purchasers
- Very high average order value
- Moderate total value driven by large, episodic purchases

These customers behave like wholesale or event-driven buyers.

**Business interpretation**
Revenue from this segment is lumpy and can introduce operational risk if not anticipated. However, each transaction is highly valuable.

**Recommended actions**
- Volume-based pricing tiers
- Clear lead-time communication
- Dedicated outreach for large orders
- Inventory planning aligned to expected bulk demand

**Goal**
Stabilize large orders while avoiding stockouts or production bottlenecks.

## Segment: Low-Value / One-Time Customers

**Profile**
- Low purchase frequency
- Low total monetary value
- Often recent but shallow engagement

This is the largest segment by count but not by revenue.

**Business interpretation**
Most customers in this segment are unlikely to become highly profitable without significant intervention. Aggressive marketing here often yields poor ROI.

**Recommended actions**
- Low-cost, automated marketing only
- Avoid discounts and incentives that erode margin
- Allow customers to self-select into higher-value segments

**Goal**
Maintain minimal engagement without allocating disproportionate resources.

## Segment: At-Risk / Churned Customers

**Profile**
- Very high recency (long time since last purchase)
- Low purchase frequency
- Low-to-moderate historical value

These customers have likely disengaged.

**Business interpretation**
While some may be reactivated, most customers in this segment are unlikely to return without significant incentives.

**Recommended actions**
- One-time reactivation campaign
- If no response, stop active marketing spend
- Do not include in core demand planning

**Goal**
Test reactivation cheaply, then move on.

## Summary

This segmentation highlights that a small subset of customers drives the majority of revenue, while most customers contribute marginally.

The primary growth levers are:
- Retaining high-value loyal customers
- Converting growing repeat customers into loyal customers
- Managing bulk buyers operationally

Treating all customers identically would result in wasted marketing spend and increased operational risk.
